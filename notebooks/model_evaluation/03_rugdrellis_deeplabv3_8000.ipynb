```python
# Ref: https://www.geeksforgeeks.org/machine-learning/how-to-plot-confusion-matrix-with-labels-in-sklearn/
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt
classes = ['sky', 'trav.', 'non-trav.', 'obstacle']
cm = np.array([
        [0.988, 0, 0.001, 0.011],
        [0, 0.948, 0.045, 0.006],
        [0.014, 0.125, 0.769, 0.092],
        [0.04,  0.026, 0.068, 0.865]
])

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classes)
disp.plot(
    cmap=plt.cm.Blues
)
# plt.xticks(rotation=45)
plt.title('SwinUperNet Normalized Confusion Matrix')
plt.show()
```
